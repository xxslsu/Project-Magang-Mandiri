# Data Preparation
menyiapkan data agar konsisten, siap uji

```python
cols_used = [
    "Tanggal Sampling",
    "Tanggal Selesai",
    "Kabupaten/Kota",
    "Kategori Sarana",
    "Komoditi",
    "Kesimpulan Uji",
    "Kesimpulan Penandaan",
    "Kesimpulan Akhir (sesuai inputan MT)"]

df = df[cols_used]
df
```

```python
df = df.rename(columns={
    "Tanggal Sampling": "tanggal_sampling",
    "Tanggal Selesai": "tanggal_hasil",
    "Kabupaten/Kota": "kab_kota",
    "Kategori Sarana": "kategori_sarana",
    "Komoditi": "komoditi",
    "Kesimpulan Uji": "kesimpulan_uji",
    "Kesimpulan Penandaan": "kesimpulan_penandaan",
    "Kesimpulan Akhir (sesuai inputan MT)": "kesimpulan_akhir"})
df.head()
```

```python
df["kesimpulan_akhir"] = df["kesimpulan_akhir"].str.upper()
df["kesimpulan_akhir"] = df["kesimpulan_akhir"].replace({
    "MEMENUHI SYARAT": "MS",
    "TIDAK MEMENUHI SYARAT": "TMS"})

df["kesimpulan_uji"] = df["kesimpulan_uji"].str.upper()
df["kesimpulan_uji"] = df["kesimpulan_uji"].replace({
    "HASIL UJI KIMIA : MS": "MS",
    "HASIL UJI KIMIA : TMS": "TMS"})

df["kesimpulan_penandaan"] = df["kesimpulan_penandaan"].str.upper()
df.head()
```

```python
bulan_id = {
    "Januari": "January",
    "Februari": "February",
    "Maret": "March",
    "April": "April",
    "Mei": "May",
    "Juni": "June",
    "Juli": "July",
    "Agustus": "August",
    "September": "September",
    "Oktober": "October",
    "November": "November",
    "Desember": "December"}
```

```python
def parse_tanggal(x):
    if pd.isna(x):
        return pd.NaT
    x = str(x).strip()
    for id_bulan, en_bulan in bulan_id.items():
        x = x.replace(id_bulan, en_bulan)
    return pd.to_datetime(x, errors="coerce", dayfirst=True)
```

```python
df["tanggal_sampling"] = df["tanggal_sampling"].apply(parse_tanggal)
df["tanggal_hasil"] = df["tanggal_hasil"].apply(parse_tanggal)
```

```python
df["lead_time"] = (df["tanggal_hasil"] - df["tanggal_sampling"]).dt.days
```

```python
data_valid = df[df["lead_time"] >= 0]
data_anomali = df[df["lead_time"] < 0]
```

```python
df_tms = df[df["kesimpulan_akhir"] == "TMS"]

penandaan_counts = (
    df_tms['kesimpulan_penandaan']
    .value_counts()
    .reindex(['MK', 'TMK'], fill_value=0))

uji_counts = (
    df_tms['kesimpulan_uji']
    .value_counts()
    .reindex(['MS', 'TMS'], fill_value=0))

data_plot = pd.DataFrame({
    'MK / MS': [penandaan_counts['MK'], uji_counts['MS']],
    'TMK / TMS': [penandaan_counts['TMK'], uji_counts['TMS']]}, index=['Penandaan', 'Uji Kimia'])

data_plot.plot(
    kind='bar',
    stacked=True,
    figsize=(7,5))

plt.title('Penyebab Produk TMS: Penandaan vs Uji Kimia')
plt.xlabel('Aspek Evaluasi')
plt.ylabel('Jumlah Produk TMS')
plt.xticks(rotation=0)
plt.legend(title='Status')
plt.tight_layout()
plt.show()
```

```python
pareto = pd.DataFrame({
    "Penandaan TMK": (df_tms["kesimpulan_penandaan"] == "TMK").sum(),
    "Uji Kimia TMS": (df_tms["kesimpulan_uji"] == "TMS").sum()}, index=["Jumlah"]).T

pareto["Persen"] = pareto["Jumlah"] / pareto["Jumlah"].sum() * 100
pareto = pareto.sort_values("Jumlah", ascending=False)
pareto
```

```python
import pandas as pd
import matplotlib.pyplot as plt

pareto["Kumulatif"] = pareto["Persen"].cumsum()

fig, ax1 = plt.subplots(figsize=(8,5))

ax1.bar(pareto.index, pareto["Jumlah"])
ax1.set_ylabel("Jumlah Kasus TMS")
ax1.set_xlabel("Penyebab TMS")

ax2 = ax1.twinx()
ax2.plot(pareto.index, pareto["Kumulatif"], marker='o')
ax2.set_ylabel("Persentase Kumulatif (%)")
ax2.set_ylim(0, 100)

ax2.axhline(80, linestyle='--')

plt.title("Pareto Chart Penyebab Produk TMS")
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency

ct_penandaan = pd.crosstab(df["kesimpulan_penandaan"], df["kesimpulan_akhir"])
chi2_pn, p_pn, _, _ = chi2_contingency(ct_penandaan)
ct_penandaan, p_pn
```

```python
ct_uji = pd.crosstab(df["kesimpulan_uji"], df["kesimpulan_akhir"])
chi2_uji, p_uji, _, _ = chi2_contingency(ct_uji)
ct_uji, p_uji
```

```python
import numpy as np
from scipy.stats import chi2_contingency

def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    r, k = confusion_matrix.shape
    return np.sqrt(chi2 / (n * (min(r-1, k-1))))

ct_penandaan = pd.crosstab(df["kesimpulan_penandaan"], df["kesimpulan_akhir"])
v_penandaan = cramers_v(ct_penandaan)

ct_uji = pd.crosstab(df["kesimpulan_uji"], df["kesimpulan_akhir"])
v_uji = cramers_v(ct_uji)
v_penandaan, v_uji
```

# seberapa besar risiko tms jika penandaan tdk sesuai

```python
ct = ct_penandaan.copy()

a = ct.loc["TMK", "TMS"] + 0.5
b = ct.loc["TMK", "MS"] + 0.5
c = ct.loc["MK", "TMS"] + 0.5
d = ct.loc["MK", "MS"] + 0.5

odds_ratio = (a * d) / (b * c)

risk_exposed = a / (a + b)
risk_unexposed = c / (c + d)
relative_risk = risk_exposed / risk_unexposed
```
